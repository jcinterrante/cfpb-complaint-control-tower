```sql
CREATE OR REPLACE TABLE cfpb_risk.silver.cfpb_complaints_clean AS
SELECT
  cr.`Complaint ID` AS complaint_id,
  TO_DATE(cr.`Date received`) AS date_received,
  DATE_TRUNC('month', TO_DATE(cr.`Date received`)) AS complaint_month,
  cr.Product AS product,
  cr.`Sub-product` AS sub_product,
  cr.Issue AS issue,
  cr.`Sub-issue` AS sub_issue,
  cr.Tags AS tags,
  cr.`Timely response?` as timely_response,
  cr.`Consumer disputed?` AS consumer_disputed,
  cr.State AS state,
  cr.`ZIP code` AS zip_code,
  bm.institution_display_name, 
  bm.rssd_id,
  bm.fdic_id,
  cr.`Consumer complaint narrative` AS consumer_complaint_narrative,
  cr.`Company public response` AS company_public_response
FROM cfpb_risk.bronze.cfpb_complaints_raw cr
JOIN cfpb_risk.reference.bank_map bm
  ON UPPER(TRIM(cr.Company)) = UPPER(TRIM(bm.company_raw));
```
