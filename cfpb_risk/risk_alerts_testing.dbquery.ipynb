```sql
--Top 20 Alerts By Bank
WITH ranked_results AS (
  SELECT
    rssd_id,
    institution_display_name,
    product,
    issue,
    risk_score,
    ROW_NUMBER() OVER (
      PARTITION BY rssd_id ORDER BY risk_score DESC) as rank
  FROM cfpb_risk.gold.risk_alerts
  GROUP BY rssd_id, institution_display_name, product, issue, risk_score
)
SELECT
  *
FROM ranked_results
WHERE rank <= 20
ORDER BY institution_display_name, rank ASC;

--Count of Alert Reasons
SELECT
  alert_reason,
  count(*) as count
FROM cfpb_risk.gold.risk_alerts
GROUP BY alert_reason
ORDER BY count DESC;

--Count of Alerts by Alert Level
SELECT
  alert_level,
  count(*) as count
FROM cfpb_risk.gold.risk_alerts
GROUP BY alert_level
ORDER BY count DESC;

-- Alert Counts by Product/Issue
SELECT
  product,
  issue,
  count(*) as count
FROM cfpb_risk.gold.risk_alerts
GROUP BY product, issue
ORDER BY count DESC;

-- Latest Month vs Prior Month
SELECT
  rssd_id, 
  institution_display_name, 
  complaint_month,
  product, 
  issue, 
  complaint_count, 
  prior_month_complaint_count
FROM cfpb_risk.gold.risk_alerts
--GROUP BY product, issue
ORDER BY complaint_count DESC
LIMIT 20;
```
