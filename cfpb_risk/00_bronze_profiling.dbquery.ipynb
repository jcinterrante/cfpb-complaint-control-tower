```sql
SELECT 
  MIN(`Date received`) AS min_date_received, 
  MAX(`Date received`) AS max_date_received
FROM cfpb_risk.bronze.cfpb_complaints_raw;

SELECT
  SUM(CASE WHEN `Complaint ID` IS NULL THEN 1 ELSE 0 END) AS null_complaint_id,
  SUM(CASE WHEN `Date received` IS NULL THEN 1 ELSE 0 END) AS null_date_received,
  SUM(CASE WHEN Company IS NULL THEN 1 ELSE 0 END) AS null_company,
  SUM(CASE WHEN Product IS NULL THEN 1 ELSE 0 END) AS null_product,
  SUM(CASE WHEN Issue IS NULL THEN 1 ELSE 0 END) AS null_issue
FROM cfpb_risk.bronze.cfpb_complaints_raw;

SELECT
  `Complaint ID`,
  COUNT(*) AS cnt
FROM cfpb_risk.bronze.cfpb_complaints_raw
GROUP BY `Complaint ID`
HAVING COUNT(*) > 1
ORDER BY cnt DESC;

SELECT
  CASE
    WHEN bm.rssd_id IS NULL THEN 'UNMATCHED'
    ELSE 'MATCHED'
  END AS match_status,
  COUNT(*) AS row_count
FROM cfpb_risk.bronze.cfpb_complaints_raw b
LEFT JOIN cfpb_risk.reference.bank_map bm
  ON UPPER(TRIM(b.Company)) = UPPER(TRIM(bm.company_raw))
GROUP BY 1;

SELECT
  b.Company,
  COUNT(*) AS row_count
FROM cfpb_risk.bronze.cfpb_complaints_raw b
LEFT JOIN cfpb_risk.reference.bank_map bm
  ON UPPER(TRIM(b.Company)) = UPPER(TRIM(bm.company_raw))
WHERE bm.rssd_id IS NULL
GROUP BY b.Company
ORDER BY row_count DESC;
```
