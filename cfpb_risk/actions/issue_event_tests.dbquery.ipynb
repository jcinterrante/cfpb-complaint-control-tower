```sql
SELECT
  *
FROM cfpb_risk.app.issues
WHERE issue_owner IS NOT NULL;

SELECT
  *
FROM cfpb_risk.app.issue_events
WHERE new_owner IS NOT NULL;

SELECT
  *
FROM cfpb_risk.app.issues
WHERE current_status != 'New';

SELECT
  *
FROM cfpb_risk.app.issue_events
WHERE new_status != 'New';

SELECT
  *
FROM cfpb_risk.app.issues
WHERE latest_note IS NOT NULL;

SELECT
  *
FROM cfpb_risk.app.issue_events
WHERE note_text IS NOT NULL;
```
