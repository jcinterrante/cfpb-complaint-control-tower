```sql
CREATE OR REPLACE VIEW cfpb_risk.app.v_issue_kpis_by_institution AS
SELECT
  institution_display_name,
  SUM(CASE WHEN coalesce(current_status, '') <> 'Closed' THEN 1 ELSE 0 END) AS open_issue_count,
  SUM(CASE WHEN current_status = 'Escalated' THEN 1 ELSE 0 END) AS escalated_issue_count,
  SUM(CASE WHEN coalesce(current_status, '') <> 'Closed' AND due_date < current_date() THEN 1 ELSE 0 END) AS overdue_issue_count,
  SUM(CASE WHEN coalesce(current_status, '') <> 'Closed' AND (issue_owner IS NULL OR trim(issue_owner) = '') THEN 1 ELSE 0 END) AS unassigned_issue_count,
  SUM(CASE WHEN coalesce(current_status, '') <> 'Closed' AND alert_level IN ('High', 'Urgent') THEN 1 ELSE 0 END) AS high_risk_open_issue_count
FROM cfpb_risk.app.issues
GROUP BY institution_display_name
ORDER BY open_issue_count DESC;
```
