```sql
CREATE OR REPLACE VIEW cfpb_risk.app.v_issue_timeline AS
SELECT
  issue_id,
  alert_id,
  institution_display_name,
  event_ts,
  event_type,
  event_user,
  old_status,
  new_status,
  old_priority,
  new_priority,
  old_owner,
  new_owner,
  old_due_date,
  new_due_date,
  note_text,
  event_comment
FROM cfpb_risk.app.issue_events
ORDER BY event_ts DESC;
```
