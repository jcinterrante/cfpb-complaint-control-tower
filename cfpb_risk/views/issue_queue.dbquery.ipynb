```sql
CREATE OR REPLACE VIEW cfpb_risk.app.v_issue_queue AS
SELECT
  issue_id,
  alert_id,
  institution_display_name,
  rssd_id,
  complaint_month,
  product,
  issue,
  risk_score,
  alert_level,
  alert_reason,
  current_status,
  priority,
  issue_owner,
  due_date,
  latest_note,
  created_ts,
  updated_ts,
  CASE 
    WHEN issue_owner IS NULL THEN 1
    ELSE 0
  END AS is_unassigned,
  CASE 
    WHEN due_date IS NOT NULL AND due_date > current_date() THEN 1
    ELSE 0
  END AS is_overdue,
  CASE
    WHEN priority = 'High' THEN 1
    WHEN priority = 'Medium' THEN 2
    WHEN priority = 'Low' THEN 3
    ELSE 9
  END AS priority_sort_order,
  CASE 
    WHEN current_status = 'New' THEN 1
    WHEN current_status = 'Escalated' THEN 2
    WHEN current_status = 'In Review' THEN 3
    WHEN current_status = 'Monitoring' THEN 4
    WHEN current_status = 'Closed' THEN 5
    ELSE 9
  END AS status_sort_order
FROM cfpb_risk.app.issues
WHERE current_status <> 'Closed'
```
